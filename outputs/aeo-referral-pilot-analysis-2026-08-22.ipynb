## tl;dr

This exploratory pilot tests whether AEO readiness is associated with discovery in neutral web-search queries. It uses four portals and eight queries; results are descriptive, not causal.

## Context & Methods

The common-core score excludes bots, rendering and structured-data dimensions because the browser environment could not verify those dimensions equivalently across all portals. Exposure is the share of eight neutral queries where a portal appeared in the top ten search results.

```python
import pandas as pd
from scipy.stats import spearmanr
pilot = pd.read_csv('aeo-referral-pilot-2026-08-22.csv')
scores = pd.DataFrame({
    'portal': ['World Bank', 'WHO Data', 'Statista', 'CEPALSTAT'],
    'common_core_score': [90.9, 83.6, 65.5, 58.2],
    'exposure': [pilot.worldbank.mean(), pilot.who.mean(), pilot.statista.mean(), pilot.cepalstat.mean()]
})
scores
```

```python
rho, p_value = spearmanr(scores['common_core_score'], scores['exposure'])
print({'spearman_rho': round(float(rho), 3), 'p_value': round(float(p_value), 3), 'n_sites': len(scores)})
```

## Results

The pilot yields rho = 0.80 with n = 4. Because the site-level sample is very small and topical fit is not randomized, the result should be presented as a directional signal only.

## Takeaways

World Bank discovery is high in this generalist query bank and aligns with its strong API/discovery score. WHO and CEPALSTAT appear mainly when the query matches their domain or geography. Statista's lack of exposure in this pilot is consistent with the query bank's official-statistics framing and should not be interpreted as a general absence from AI answers.
